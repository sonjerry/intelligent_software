```python
# 큐(Queue)를 효율적으로 구현할 수 있는 도구
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        v = queue.popleft()
        print(v, end=' ')
        for neighbor in graph[v]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

print("\nBFS 탐색 순서:")
bfs(graph, '이가을')
```

```python
# 사용 Ai: Codex
# 프롬프트:  bfs, dfs, 친구추천 시스템을 networkx 로 시각화,
#          테스트 케이스를 최소 3개 작성한다.
#          - 하: 노드 10개 내외, 최대 깊이 2
#          - 중: 노드 100개 내외, 최대 깊이 10
#          - 상: 노드 1,000개 내외, 최대 깊이 50

%pip -q install koreanize-matplotlib
```

```python
from collections import deque

import koreanize_matplotlib
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.colors import Normalize


def bfs(graph, start):
    queue = deque([start])
    order = []
    distance = {start: 0}

    while queue:
        person = queue.popleft()
        order.append(person)

        for friend in graph[person]:
            if friend not in distance:
                distance[friend] = distance[person] + 1
                queue.append(friend)

    return order, distance


def dfs(graph, start):
    visited = set()
    order = []

    def visit(person):
        visited.add(person)
        order.append(person)

        for friend in graph[person]:
            if friend not in visited:
                visit(friend)

    visit(start)
    return order


def recommend_friend(bfs_order, distance):
    return [person for person in bfs_order if distance[person] == 2]


def make_graph(node_count, max_depth):
    graph = {'Alice': []}
    previous_level = ['Alice']
    next_id = 1
    nodes_per_level, extra_nodes = divmod(node_count - 1, max_depth)

    for depth in range(1, max_depth + 1):
        level_size = nodes_per_level + int(depth <= extra_nodes)
        current_level = [
            f'N{node_id:04d}'
            for node_id in range(next_id, next_id + level_size)
        ]

        for index, person in enumerate(current_level):
            parent_index = index * len(previous_level) // level_size
            graph[previous_level[parent_index]].append(person)
            graph[person] = []

        previous_level = current_level
        next_id += level_size

    return graph


def visualize(graph, start, difficulty, expected_nodes, expected_depth):
    bfs_order, distance = bfs(graph, start)
    dfs_order = dfs(graph, start)
    recommendations = recommend_friend(bfs_order, distance)

    assert len(graph) == expected_nodes
    assert len(bfs_order) == len(dfs_order) == expected_nodes
    assert max(distance.values()) == expected_depth

    G = nx.DiGraph()
    G.add_nodes_from(graph)
    G.add_edges_from(
        (person, friend)
        for person, friends in graph.items()
        for friend in friends
    )

    levels = {}
    for person in bfs_order:
        levels.setdefault(distance[person], []).append(person)

    pos = {}
    for depth, people in levels.items():
        for index, person in enumerate(people):
            y = 0.5 if len(people) == 1 else 1 - index / (len(people) - 1)
            pos[person] = (depth, y)

    small_graph = expected_nodes <= 10
    edge_options = {'arrowsize': 16} if small_graph else {}
    node_size = 1150 if small_graph else 85 if expected_nodes <= 100 else 18

    fig, axes = plt.subplots(3, 1, figsize=(19, 11), constrained_layout=True)
    fig.suptitle(f'{difficulty} · 노드 {len(G)}개 · 최대 깊이 {max(distance.values())}')

    for ax, title, order in [
        (axes[0], 'BFS 탐색 순서', bfs_order),
        (axes[1], 'DFS 탐색 순서', dfs_order),
    ]:
        ranks = {
            person: rank
            for rank, person in enumerate(order, start=1)
        }

        nx.draw_networkx_edges(
            G, pos, ax=ax, arrows=small_graph,
            node_size=node_size, edge_color='#94A3B8', alpha=0.55,
            **edge_options,
        )
        nx.draw_networkx_nodes(
            G, pos, ax=ax, nodelist=bfs_order, node_size=node_size,
            node_color=[ranks[person] for person in bfs_order],
            cmap=plt.cm.YlGnBu, vmin=1, vmax=len(order),
        )

        if small_graph:
            for person in bfs_order:
                x, y = pos[person]
                color = 'white' if ranks[person] > len(order) * 0.65 else '#111827'
                ax.text(
                    x, y, f'{ranks[person]}\n{person}',
                    ha='center', va='center', fontsize=9, color=color,
                )
        else:
            scale = plt.cm.ScalarMappable(
                norm=Normalize(1, len(order)), cmap=plt.cm.YlGnBu,
            )
            fig.colorbar(scale, ax=ax, label='방문 순번', pad=0.01)

        ax.set_title(title)

    ax = axes[2]
    recommendation_set = set(recommendations)

    nx.draw_networkx_edges(
        G, pos, ax=ax, arrows=small_graph,
        node_size=node_size, edge_color='#94A3B8', alpha=0.4,
        **edge_options,
    )
    nx.draw_networkx_nodes(
        G, pos, ax=ax, nodelist=bfs_order, node_size=node_size,
        node_color=[
            '#FBBF24' if person == start else
            '#22C55E' if person in recommendation_set else '#CBD5E1'
            for person in bfs_order
        ],
    )

    if small_graph:
        nx.draw_networkx_labels(G, pos, ax=ax, font_size=9)

    ax.set_title(f'친구 추천 · {start}에서 2단계 거리 · {len(recommendations)}명')

    for ax in axes:
        ax.set_xlim(-0.15, expected_depth + 0.15)
        ax.set_ylim(-0.12, 1.12)
        ax.set_axis_off()

    print(f'[{difficulty}] 노드 {len(G)}개, 최대 깊이 {max(distance.values())}')
    for title, order in [('BFS', bfs_order), ('DFS', dfs_order)]:
        limit = 20
        suffix = f' … (총 {len(order)}개)' if len(order) > limit else ''
        print(f'{title}: {" → ".join(order[:limit])}{suffix}')

    print('친구 추천:', ', '.join(recommendations))
    plt.show()


small_graph = {
    'Alice': ['안우현', '이가을', '정민서'],
    '안우현': ['최우진', '배정우'],
    '이가을': ['김준석', '강성민'],
    '정민서': ['이준수', '김준성'],
    '최우진': [],
    '배정우': [],
    '김준석': [],
    '강성민': [],
    '이준수': [],
    '김준성': [],
}

test_cases = [
    ('하', small_graph, 10, 2),
    ('중', make_graph(100, 10), 100, 10),
    ('상', make_graph(1000, 50), 1000, 50),
]

for difficulty, graph, node_count, max_depth in test_cases:
    visualize(graph, 'Alice', difficulty, node_count, max_depth)
```

```python
# 사용 AI: Codex
# 프롬프트: 감염 시뮬레이션으로 확장하여 2단계 이내 전염 경로를 추적한다.
#          - 그래프는 인접 리스트 형태로 구성한다.
#          - 최초 감염자는 1명이며, 이름으로 지정한다. (예: 'Alice')
#          - 최초 감염자를 제외하고 2단계 이내로 연결된 노드를 출력한다.
#          - 출력 형식은 자유이며, 각 노드까지의 감염 경로와 거리(레벨)를 함께 표시한다.
#          테스트 케이스를 최소 3개 작성한다.
#          - 하: 노드 10개 내외, 최대 깊이 2
#          - 중: 노드 100개 내외, 최대 깊이 10
#          - 상: 노드 1,000개 내외, 최대 깊이 50

%pip -q install networkx koreanize-matplotlib
```

```python

from collections import deque

import koreanize_matplotlib
import matplotlib.pyplot as plt
import networkx as nx

FONT = 'NanumGothic'
plt.rcParams['font.family'] = FONT


def make_graph(node_count, max_depth):
    graph = {'Alice': []}
    previous_level = ['Alice']
    next_id = 1
    nodes_per_level, extra_nodes = divmod(node_count - 1, max_depth)

    for depth in range(1, max_depth + 1):
        level_size = nodes_per_level + int(depth <= extra_nodes)
        current_level = [
            f'N{node_id:04d}'
            for node_id in range(next_id, next_id + level_size)
        ]

        for index, person in enumerate(current_level):
            parent_index = index * len(previous_level) // level_size
            graph[previous_level[parent_index]].append(person)
            graph[person] = []

        previous_level = current_level
        next_id += level_size

    return graph


def trace_infection(graph, infected):
    if infected not in graph:
        raise ValueError(f'그래프에 {infected} 노드가 없습니다.')

    queue = deque([infected])
    distance = {infected: 0}
    paths = {infected: (infected,)}
    results = []

    while queue:
        person = queue.popleft()
        depth = distance[person]

        if depth == 2:
            continue

        for neighbor in graph[person]:
            if neighbor in distance:
                continue

            distance[neighbor] = depth + 1
            paths[neighbor] = (*paths[person], neighbor)
            results.append((neighbor, distance[neighbor], paths[neighbor]))
            queue.append(neighbor)

    return results


def visualize_infection(infected, results, difficulty):
    network = nx.DiGraph()
    network.add_node(infected, level=0)

    levels = {0: [infected], 1: [], 2: []}

    for person, depth, path in results:
        network.add_node(person, level=depth)
        network.add_edge(path[-2], person)
        levels[depth].append(person)

    positions = {}
    for depth, people in levels.items():
        for index, person in enumerate(people):
            y = 0.5 if len(people) == 1 else 1 - index / (len(people) - 1)
            positions[person] = (depth, y)

    colors = {0: '#EF4444', 1: '#FB923C', 2: '#FDE68A'}
    show_labels = len(network) <= 10
    node_size = 1100 if show_labels else 100

    fig, ax = plt.subplots(
        figsize=(10, 5 if show_labels else 8),
        constrained_layout=True,
    )

    nx.draw_networkx_edges(
        network, positions, ax=ax, arrows=True,
        arrowsize=11, node_size=node_size, edge_color='#94A3B8',
    )
    nx.draw_networkx_nodes(
        network, positions, ax=ax, node_size=node_size,
        node_color=[colors[network.nodes[person]['level']] for person in network],
    )

    if show_labels:
        nx.draw_networkx_labels(
            network, positions, ax=ax, font_size=9, font_family=FONT,
        )

    ax.set_title(f'[{difficulty}] {infected}의 2단계 이내 전염 경로')
    ax.set_xticks([0, 1, 2], ['최초 감염자', '1단계', '2단계'])
    ax.tick_params(axis='x', bottom=False, labelbottom=True)
    ax.set_yticks([])
    ax.set_xlim(-0.25, 2.25)
    ax.set_ylim(-0.1, 1.1)

    for spine in ax.spines.values():
        spine.set_visible(False)

    plt.show()


small_graph = {
    'Alice': ['안우현', '이가을', '정민서'],
    '안우현': ['최우진', '배정우'],
    '이가을': ['김준석', '강성민'],
    '정민서': ['이준수', '김준성'],
    '최우진': [],
    '배정우': [],
    '김준석': [],
    '강성민': [],
    '이준수': [],
    '김준성': [],
}

test_cases = [
    ('하', small_graph, 10, 2),
    ('중', make_graph(100, 10), 100, 10),
    ('상', make_graph(1000, 50), 1000, 50),
]

infected = 'Alice'

for difficulty, graph, node_count, max_depth in test_cases:
    results = trace_infection(graph, infected)

    network = nx.DiGraph(graph)
    expected = nx.single_source_shortest_path_length(network, infected)
    expected_within_two = {
        person: depth for person, depth in expected.items() if 1 <= depth <= 2
    }

    assert len(network) == len(expected) == node_count
    assert max(expected.values()) == max_depth
    assert len(results) == len(expected_within_two)
    assert {person: depth for person, depth, _ in results} == expected_within_two

    print(f'\n[{difficulty}] 노드 {node_count}개 / 최대 깊이 {max_depth}')
    print(f'최초 감염자: {infected} / 2단계 이내 도달: {len(results)}명')

    for person, depth, path in results:
        print(f'{depth}단계 | {person} | {" → ".join(path)}')

    visualize_infection(infected, results, difficulty)
```
